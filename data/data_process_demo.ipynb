```python
import random
import copy
```

```python
def build_conversation(data: dict, id: int) -> dict:
    conversation_dict: dict = {}
    conversation_dict['id'] = id
    conversation_dict['task'] = "订酒店"
    conversation_dict['extract_slot'] = 1

    conversation_list = []
    conversation_state = copy.deepcopy(empty_keys_dict)
    # print('conversation state: ', conversation_state)
    slot = ''
    for i in range(len(key_list) + 1):
        single_turn_dict = {}
        single_turn_dict['Turn'] = i
        single_turn_dict['origin_slots'] = copy.deepcopy(conversation_state)
        single_conversation = []
        
        llm_query = ""
        user_ans_value = ""
        user_ans_str = ""
        
        user_ans_value = (data[slot] if type(data[slot]) != list else random.choice(data[slot])) if i != 0 else ""
        user_ans_str = (random.choice(answer_template_list[slot]).format(user_ans_value)) if i != 0 else ""
        
        user_dict = {
            'from': 'user',
            'value': user_ans_str
        }
        if slot != '' : conversation_state[slot] = user_ans_value 

        # 抽取新的slot
        slot = random.choice([key for key, value in conversation_state.items() if value is None]) if i != len(key_list) else None
        llm_query += random.choice(query_template_list[slot]) if i != len(key_list) else ""
        # print(llm_query)
        llm_dict = {
            'from': 'assistant',
            'value': llm_query if i != len(key_list) else "我已获得到所有的信息，以下是信息内容：" + json.dumps(conversation_state, ensure_ascii=False) 
        }
        single_conversation.append(user_dict)
        single_conversation.append(llm_dict)
        single_turn_dict['conversations'] = single_conversation
        single_turn_dict['new_slots'] = copy.deepcopy(conversation_state)
        
        conversation_list.append(single_turn_dict)

    
    
    conversation_dict['content'] = conversation_list
    return conversation_dict
```

## 扰动混入

句子级：答非所问，主题不合理不相干；

每场完整对话中，用户方有一定概率犯错误（至多一次）；

含有错误的对话占总数据量5%；

```python
import json
import random
from copy import deepcopy
from tqdm import tqdm
```

```python
re_ask_templete_list = [
    "不好意思，您的回答不符合要求。",
    "对不起，您可能理解错我的意思。",
    "对不起，您提供的信息有误。",
    "不好意思，您的回答不合理，请您重新提供信息。",
    "不好意思，您的回答不符合要求，请您重新回答。"
]
```

```python
def _add_noise(data: dict) -> dict:
    content = data['content'].copy()
    wrong_data_idx = random.choice(range(1, len(content) - 1))
    wrong_turn = deepcopy(content[wrong_data_idx])

    # 修改该block
    wrong_turn['origin_slots'] = wrong_turn['new_slots']
    last_ask = wrong_turn['conversations'][1]['value']
    wrong_turn['conversations'][1]['value'] = random.choice(re_ask_templete_list) + last_ask

    content.insert(wrong_data_idx + 1, wrong_turn)

    # turn重新赋值
    for i in range(len(content)): content[i]['Turn'] = i

    data['content'] = content
    return data

def add_noise(data_list, noise_rate=0.05) -> list:
    new_data_list = []
    for data in tqdm(data_list):
        if random.random() < noise_rate:
            noise_data = _add_noise(data)
            new_data_list.append(noise_data)
        else:
            new_data_list.append(data)
    return new_data_list
```

```python
with open('./output/conversation_hotel.json', 'r', encoding='utf8') as input_file:
    data_list = json.load(input_file)
new_data_list: list = add_noise(data_list, 0.05)
with open('./output/conversation_hotel_noise.json', 'w', encoding='utf8') as output_file:
    json.dump(new_data_list, output_file, ensure_ascii=False, indent=4)
```
